## 10. Actividad - Operadores unarios

```python
variacion_peso = 5
print("Unario positivo:", +variacion_peso)
print("Unario negativo:", -variacion_peso)
temperatura = 3.5
print("\nTemperatura:", temperatura)
print("Valor positivo:", +temperatura)
print("Valor negativo:", -temperatura)
```

Ahora cambia temperatura a -2.5 y explica qué ocurre al aplicar +temperatura y -temperatura.

```python
temperatura = -2.5
print(+temperatura)
print(-temperatura)
```

- Lo que ocurre al usar operadores unarios es que, el menos unario (-) invierte el signo del valor.
- Si es un número positivo, lo vuelve negativo, y si es negativo, lo vuelve positivo.
- Por otro lado, el más unario (+) mantiene el valor sin cambios.

## 15. Actividad - División entera

```python
numero_muestras = 17
grupos = 5
division_normal = numero_muestras / grupos
division_entera = numero_muestras // grupos
print("División normal:", division_normal)
print("División entera:", division_entera)
```

Contesta: ¿Qué diferencia observas entre / y // ?
- En la división normal (/), existen los decimales, en la división entera (//) solo se toma la parte entera, no hay decimales, también, redondea hacia abajo (hacia el entero más pequeño).
