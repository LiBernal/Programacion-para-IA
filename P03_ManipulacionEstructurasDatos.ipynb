## 10. Actividad - Capturar una nueva observación desde teclado

Registra una observación solicitando los datos al usuario.


```python
nueva_muestra = {
    "id": int(input("ID: ")),
    "nombre": input("Clase de fruta: "),
    "peso": float(input("Peso en gramos: ")),
    "diametro": float(input("Diámetro en cm: ")),
    "dulzura": int(input("Dulzura (0-10): ")),
    "acidez": int(input("Acidez (0-10): "))
}

dataset.append(nueva_muestra)

print("\nRegistro agregado:")
print(nueva_muestra)

print("\nCantidad total de observaciones:", len(dataset))
```

## 12. Actividad - Guardar el dataset en un archivo

El material del tema incluye escritura de información en archivos. A continuación se guardará el dataset en un archivo de texto.


```python
archivo = open("dataset.txt", "w", encoding="utf-8")
archivo.write(str(dataset))
archivo.close()

print("Dataset guardado en dataset.txt")
```

### Verifica

Busca el archivo `dataset.txt` en la misma carpeta del notebook y revisa su contenido.


## 13. Actividad - Leer el archivo

Ahora abre el archivo en modo lectura.


```python
archivo = open("dataset.txt", "r", encoding="utf-8")
contenido = archivo.read()
archivo.close()

print("Contenido del archivo:")
print(contenido)
```
