## 7. Actividad  -  Estructura `if-elif-else`

Utiliza varias condiciones para clasificar un nivel de dulzura.


````python
dulzura = 8

if dulzura >= 8:
    nivel = "Alta"
elif dulzura >= 5:
    nivel = "Media"
else:
    nivel = "Baja"

print("Nivel de dulzura:", nivel)
````

- 5 = Nivel de dulzura: "Media"
- 3 = Nivel de dulzura: "Baja"
- 1 = Nivel de dulzura: "Baja"

¿Por qué el orden de las condiciones es importante?
- El orden de las condiciones es importante, ya que determinan como se van evaluando, en Python se evaluan de abajo hacia arriba, y se ejecuta la que encuentre como verdadera primero.

## 10. Actividad  -  Bucle `for`

Un bucle `for` permite recorrer una secuencia.


````python
pesos = [185, 210, 95, 170, 120]

for peso in pesos:
    print("Procesando peso:", peso)
````

### Analiza

1. ¿Cuántas veces se ejecutó el `print`?
- Se ejecutó 5 veces.
2. ¿Qué valor toma la variable `peso` en cada iteración?
- Va tomando los valores: 185, 210, 95, 170, 120.
3. ¿Qué estructura se está recorriendo?
- Se está recorriendo una lista, que se llama pesos.


## 11. Actividad  -  Acumulador con `for`

Calcula la suma y el promedio de los pesos.


````python
pesos = [185, 210, 95, 170, 120]

suma = 0

for peso in pesos:
    suma += peso

promedio = suma / len(pesos)

print("Suma de pesos:", suma)
print("Peso promedio:", promedio)
````

### Responde

1. ¿Qué función cumple la variable `suma`?
- La variable suma se usa como un acumulador, que va recibiendo los valores de la lista de pesos y los va guardando y sumando.
2. ¿Por qué debe iniciar en `0`?
- Se inicializa en 0 porque aún no debería tener ningún valor acumulado.
3. ¿Qué función cumple `len(pesos)`?
- len es length, que significa el largo, y en python es la manera de obtener el numero de elementos que tiene la lista, aquí se está usando para calcular el promedio.


## 12. Actividad  -  Condicional dentro de un `for`

Combina repetición y decisión.


````python
dulzuras = [8, 6, 2, 9, 4]

for valor in dulzuras:
    if valor >= 7:
        print(valor, "-> Dulzura alta")
    elif valor >= 5:
        print(valor, "-> Dulzura media")
    else:
        print(valor, "-> Dulzura baja")
````

### Actividad

Modifica la lista para incluir al menos 8 valores y observa la clasificación de cada uno.


````python
# Modifica la lista y vuelve a ejecutar
dulzuras = [8, 6, 2, 9, 4, 7, 5, 1]

for valor in dulzuras:
    if valor >= 7:
        print(valor, "-> Dulzura alta")
    elif valor >= 5:
        print(valor, "-> Dulzura media")
    else:
        print(valor, "-> Dulzura baja")
````

## 13. Actividad  -  Bucle `while`

Un ciclo `while` se ejecuta mientras una condición sea verdadera.


````python
contador = 1

while contador <= 5:
    print("Iteración:", contador)
    contador += 1

print("Fin del ciclo")
````

### Analiza

1. ¿Cuál es la condición del ciclo?
- Mientras el valor de la variable contador sea menor o igual que 5, se seguirá repitiendo.
2. ¿Qué variable controla las repeticiones?
- La variable contador.
3. ¿Qué ocurriría si se elimina `contador += 1`?
- Se seguiría repitiendo el ciclo, ya que, no hay una función que le diga que pare.


## 14. Actividad  -  Menú repetitivo con `while`

Crea un menú sencillo.


````python
opcion = ""

while opcion != "salir":

    print("\nMENÚ")
    print("Escribe 'capturar' para registrar un peso.")
    print("Escribe 'salir' para terminar.")

    opcion = input("Opción: ").lower()

    if opcion == "capturar":
        peso = float(input("Peso en gramos: "))
        print("Dato registrado:", peso)

    elif opcion == "salir":
        print("Finalizando programa...")

    else:
        print("Opción no válida.")
````

## 15. Actividad  -  Procesamiento de un pequeño dataset

Combina estructuras de datos, `for` y condicionales.


````python
dataset = [
    {"nombre": "Manzana", "dulzura": 8, "acidez": 4},
    {"nombre": "Naranja", "dulzura": 7, "acidez": 6},
    {"nombre": "Limón", "dulzura": 2, "acidez": 9},
    {"nombre": "Pera", "dulzura": 8, "acidez": 3}
]

for fruta in dataset:

    print("\nFruta real:", fruta["nombre"])

    if fruta["dulzura"] >= 7 and fruta["acidez"] <= 5:
        prediccion = "Manzana"

    elif fruta["dulzura"] >= 5 and fruta["acidez"] <= 7:
        prediccion = "Naranja"

    else:
        prediccion = "Limón"

    print("Predicción por reglas:", prediccion)
````
