## 5. Actividad -. Listas

Una lista permite almacenar varios elementos en una colección ordenada y modificable.

Ejecuta el siguiente ejemplo:


```python
pesos = [185.5, 210.0, 95.0, 160.5, 120.0]

print("Lista completa:")
print(pesos)

print("\nPrimera muestra:", pesos[0])
print("Segunda muestra:", pesos[1])
print("Última muestra:", pesos[-1])
print("Cantidad de elementos:", len(pesos))
```

### Analiza

Completa la tabla:

| Expresión | Resultado |
|---|---|
| `pesos[0]` | 185.5 |
| `pesos[1]` | 210.0 |
| `pesos[-1]` | 120.0 |
| `len(pesos)` | 5 |

### Responde

1. ¿Por qué una lista es adecuada para almacenar varios pesos?
- Una lista permite valores duplicados y puede ser modificable, es adecuada para almacenar variables mutables.
2. ¿Qué representa el índice `0`?
- El primer elemento de la lista.
3. ¿Qué significa utilizar el índice `-1`?
- El ultimo elemento de la lista.

## 6. Actividad - Modificación de una lista

Las listas son **mutables**, por lo que sus elementos pueden modificarse.


```python
pesos = [185.5, 210.0, 95.0]

print("Antes:", pesos)

pesos[0] = 190.0

print("Después:", pesos)
```

### Experimenta

Modifica el segundo y tercer elemento de la lista y ejecuta nuevamente la celda.


```python
# Modifica los valores y observa el resultado

pesos = [190.0, 210.0, 95.0]

# Cambia aquí el segundo y tercer elemento
pesos[1] = 4.00
pesos[2] = 16.00

print(pesos)
```

## 7. Actividad - Tuplas

Una tupla es una colección ordenada pero **inmutable**.

En este ejemplo se utilizará una tupla para representar un color RGB.


```python
color_manzana = (220, 30, 40)

print("Color RGB:", color_manzana)
print("Rojo:", color_manzana[0])
print("Verde:", color_manzana[1])
print("Azul:", color_manzana[2])
```

### Analiza

1. ¿Cuántos elementos contiene la tupla?
- 3 elementos.
2. ¿Qué representa cada posición?
- En este caso representa un color, el índice de color_manzana es el que representa de que color se trata.
3. ¿Por qué podría ser conveniente usar una tupla cuando los datos no deben modificarse?
- Nos ayuda si queremos valores constantes, pueden evitar equivocacciones, o pueden servir como seguridad incluso.

### Prueba de inmutabilidad

La siguiente instrucción produciría un error porque una tupla no puede modificarse directamente.

Descomenta la línea únicamente para observar el error.


```python
color_manzana = (220, 30, 40)

# Descomenta para experimentar:
color_manzana[0] = 255
```

## 8. Actividad - Conjuntos

Un conjunto almacena elementos únicos y elimina duplicados.


```python
clases = {"Manzana", "Naranja", "Limón", "Manzana", "Naranja"}

print("Clases registradas:")
print(clases)

print("Cantidad de clases únicas:", len(clases))
```

### Responde

1. ¿Por qué no aparecen repetidas `Manzana` y `Naranja`?
- Porque en un conjunto, ambas pertenecen a una clase o tipo de característica, y no se permiten los duplicados.
2. ¿Qué utilidad podría tener un conjunto para conocer las clases disponibles en un dataset?
- En un dataset es común que se repitan características o clases, un conjunto puede recorrer todos los registros del dataset y decir exactamente que clases existen sin importar cuantas veces se repitan.
3. ¿Puede accederse a los elementos de un conjunto mediante un índice como `clases[0]`?
- No se puede, son colecciones desordenadas.

## 9. Actividad - Diccionarios

Un diccionario almacena información mediante pares **clave-valor**.

En el siguiente ejemplo se representa una observación de una fruta.


```python
fruta = {
    "nombre": "Manzana",
    "peso": 185.5,
    "diametro": 7.3,
    "dulzura": 8,
    "acidez": 4
}

print(fruta)

print("\nNombre:", fruta["nombre"])
print("Peso:", fruta["peso"])
print("Diámetro:", fruta["diametro"])
print("Dulzura:", fruta["dulzura"])
print("Acidez:", fruta["acidez"])
```

### Completa

| Clave | Valor |
|---|---|
| `nombre` | Manzana |
| `peso` | 185.5 |
| `diametro` | 7.3 |
| `dulzura` | 8 |
| `acidez` | 4 |

### Reflexiona

¿Por qué un diccionario es adecuado para representar una observación?
- Una observación puede ser una pequeña nota, como una característica, y la forma en que se mide como resultado, un diccionario, al ser clave-valor, permite almacenar diferentes datos facilmente.



## 10. Actividad - Modificación de un diccionario

Los diccionarios pueden modificarse.


```python
fruta = {
    "nombre": "Manzana",
    "peso": 185.5,
    "diametro": 7.3,
    "dulzura": 8,
    "acidez": 4
}

print("Antes:")
print(fruta)

fruta["peso"] = 190.0
fruta["color"] = "Rojo"

print("\nDespués:")
print(fruta)
```

### Analiza

1. ¿Qué clave fue modificada?
- peso
2. ¿Qué nueva clave fue agregada?
- color
3. ¿Qué diferencia existe entre modificar una clave existente y agregar una nueva?
- Al modificar una clave existente, se modifica normalmente el valor que acompaña a esa clave, al agregar una clave, se tiene que agregar ambos, la clave y su valor correspondiente.

## 11. Actividad - Estructuras anidadas

Las estructuras anidadas permiten combinar listas, tuplas, diccionarios y conjuntos.

En este ejemplo se creará una **lista de diccionarios**.


```python
dataset = [
    {
        "nombre": "Manzana",
        "peso": 185.5,
        "diametro": 7.3,
        "dulzura": 8,
        "acidez": 4
    },
    {
        "nombre": "Naranja",
        "peso": 210.0,
        "diametro": 8.1,
        "dulzura": 7,
        "acidez": 6
    },
    {
        "nombre": "Limón",
        "peso": 95.0,
        "diametro": 5.4,
        "dulzura": 2,
        "acidez": 9
    }
]

print(dataset)
```

### Observa la estructura

En este caso:

- la estructura externa es una **lista**;
- cada elemento de la lista es un **diccionario**;
- cada diccionario representa una **observación**;
- las claves `peso`, `diametro`, `dulzura` y `acidez` representan características;
- la clave `nombre` representa la clase de la observación.


## 12. Actividad - Acceso a información dentro de estructuras anidadas

Ejecuta:


```python
print("Primera observación:")
print(dataset[0])

print("\nNombre de la primera observación:")
print(dataset[0]["nombre"])

print("\nPeso de la segunda observación:")
print(dataset[1]["peso"])

print("\nAcidez de la tercera observación:")
print(dataset[2]["acidez"])
```

### Completa

| Expresión | Resultado |
|---|---|
| `dataset[0]` | {'nombre': 'Manzana', 'peso': 185.5, 'diametro': 7.3, 'dulzura': 8, 'acidez': 4}
 |
| `dataset[0]["nombre"]` | Manzana |
| `dataset[1]["peso"]` | 210.0 |
| `dataset[2]["acidez"]` | 9 |

### Responde

¿Qué ocurre primero en la expresión `dataset[1]["peso"]`: el acceso a la lista o el acceso al diccionario?
- Primero ocurre el acceso a la lista mediante el índice (en este caso 1), después se accede al diccionario, mediante su valor clave.

## 18. Análisis

Responde con tus propias palabras:

1. ¿Qué es una lista?
- Es una estructura de datos, mutable, que permite almacenar elementos, los cuales se pueden repetir.
2. ¿Qué significa que una lista sea mutable?
- Significa que sus elementos se pueden modificar, agregar o eliminar directamente en memoria sin necesidad de crear una lista nueva.
3. ¿Qué diferencia existe entre una lista y una tupla?
- Las listas son mutables (se pueden editar) y usan corchetes [], mientras que las tuplas son inmutables y usan paréntesis ().
4. ¿Qué característica principal tiene un conjunto?
- Es una colección no ordenada de elementos donde no se permiten duplicados.
5. ¿Por qué un conjunto puede ser útil al analizar clases de un dataset?
- Porque permite identificar rápidamente cuántas y cuáles son todas las categorías o clases diferentes, eliminando repetidos.
6. ¿Qué es un diccionario?
- Es una estructura de datos que almacena elementos en forma de parejas clave-valor.
7. ¿Qué diferencia existe entre un índice y una clave?
- Un índice es un número entero de posición dentro de una secuencia (como en listas o tuplas), mientras que una clave es un identificador (generalmente un texto o número) usado para acceder a un valor en un diccionario.
8. ¿Qué es una estructura anidada?
- Es una estructura de datos que contiene otras estructuras dentro de sí misma, como una lista de diccionarios o un diccionario con tuplas y conjuntos.
9. ¿Por qué una lista de diccionarios puede representar un dataset?
- Porque la lista representa la colección completa de registros (filas), mientras que cada diccionario representa una observación individual vinculando los nombres de los atributos (claves) con sus valores.
10. ¿Qué representa una observación?
- Representa un elemento o instancia individual de estudio en un conjunto de datos (por ejemplo, una fruta específica o un vehículo).
11. ¿Qué representa una característica?
- Una propiedad, atributo o variable de la observación que aporta información descriptiva (por ejemplo, el peso, diámetro o velocidad).
12. ¿Qué representa una clase?
- La categoría o etiqueta asignada a la observación para resolver un problema de clasificación (por ejemplo, el tipo de fruta o modelo).
13. ¿Qué estructura utilizarías para almacenar colores RGB que no deban cambiar?
- Una tupla
14. ¿Qué estructura utilizarías para eliminar valores duplicados?
- Un conjunto
15. ¿Qué estructura utilizarías para almacenar registros con nombre, peso y diámetro?
- Un diccionario.
